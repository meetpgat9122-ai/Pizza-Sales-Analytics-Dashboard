# Pizza Sales Analytics — Data Exploration

## Objective

The objective of this notebook is to understand, profile,
validate, and prepare the pizza sales dataset before performing
SQL analysis and building the Power BI dashboard.

## Dataset

The dataset contains pizza order transaction records with
information about orders, quantities, prices, pizza sizes,
categories, ingredients, and pizza names.

```python
import pandas as pd

def convert_date(x):
    x = str(x).strip()
    
    if "-" in x:
        return pd.to_datetime(x, format="%d-%m-%Y")
    else:
        return pd.to_datetime(x, format="%m/%d/%Y")

df["order_date"] = df["order_date"].apply(convert_date)
```

```python
df["calculated_total"] = df["quantity"] * df["unit_price"]

df["difference"] = (
    df["total_price"] - df["calculated_total"]
)
```

```python
df["year"] = df["order_date"].dt.year
df["month_number"] = df["order_date"].dt.month
df["month_name"] = df["order_date"].dt.month_name()
df["day_name"] = df["order_date"].dt.day_name()
```

```python
df["order_time"].head()
```

```python
df["order_time"] = pd.to_datetime(
    df["order_time"],
    format="%H:%M:%S"
)
```

```python
df["order_hour"] = df["order_time"].dt.hour
```

```python
df.drop(
    columns=["calculated_total", "difference"],
    inplace=True
)
```
